## Import Dependancies

```python
import import_ipynb
data = __import__("01_data_exploration")
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
numeric_cols = [
    "rainfall_7d_mm", "monthly_rainfall_mm", "elevation_m", "distance_to_river_m",
    "drainage_index", "ndvi", "ndwi", "historical_flood_count", "infrastructure_score",
    "population_density_per_km2", "built_up_percent", "nearest_hospital_km",
    "nearest_evac_km", "flood_risk_score",
]

corr = data.df[numeric_cols].corr()["flood_risk_score"].drop("flood_risk_score").sort_values()

fig, ax = plt.subplots(figsize=(7, 5))
corr.plot(kind="barh", ax=ax, color=["#dc3545" if v < 0 else "#0d6efd" for v in corr])
ax.set_title("Correlation with flood_risk_score")
ax.set_xlabel("Pearson correlation")
ax.axvline(0, color="black", linewidth=1)
ax.xaxis.grid(True, linestyle="--", alpha=0.6)
ax.yaxis.grid(True, linestyle="--", alpha=0.4)
ax.set_axisbelow(True)
plt.show()
```
